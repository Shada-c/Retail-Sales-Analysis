```python
import pandas as pd
df = pd.read_csv('/content/Payments.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df['Payment_Method'].unique()
```

```python
df['Payment_Method'] = df['Payment_Method'].str.strip()
df['Payment_Method'] = df['Payment_Method'].str.capitalize()
```

```python
df['Payment_Method'] = df['Payment_Method'].str.replace('Cashh','Cash')
df['Payment_Method'] = df['Payment_Method'].str.replace('?','')
df['Payment_Method'] = df['Payment_Method'].str.replace('Upi-unknown','UPI')
df['Payment_Method'] = df['Payment_Method'].str.replace('Upi','UPI')
```

```python
df['Payment_Status'] = df['Payment_Status'].str.strip()
df['Payment_Status'] = df['Payment_Status'].str.capitalize()
```

```python
df['Payment_Status'].unique()
```

```python
df['Payment_Date'].unique()
```

```python
df['Payment_Date'] = pd.to_datetime(df['Payment_Date'], format='mixed', dayfirst=True, errors='coerce')
```

```python
df['Payment_Date'] = df['Payment_Date'].dt.strftime('%d-%m-%Y')
```

```python
df['Amount_Paid'].unique()
```

```python
df = df.drop('Amount_Paid',axis= 1)
```

```python
df.duplicated().sum()
```

```python
df= df.drop_duplicates()
```
