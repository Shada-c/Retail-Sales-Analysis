```python
import pandas as pd
df = pd.read_csv('/content/Products.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['Product_Name'].unique()
```

```python
df['Product_Name'] = df['Product_Name'].str.strip()
```

```python
df['Category'].unique()
```

```python
df['Category'] = df['Category'].str.strip()
df['Category'] = df['Category'].str.capitalize()
```

```python
df['Category'] = df['Category'].str.strip()
```

```python
df['Category'].unique()
```

```python
df['Category'] = df['Category'].replace('Electronic','Electronics')
```

```python
df['Category'] = df['Category'].replace('Electornics','Electronics')
```

```python
df['Subcategory'].unique()
```

```python
df['Subcategory'] = df['Subcategory'].str.strip()
df['Subcategory'] = df['Subcategory'].str.capitalize()
```

```python
df['Subcategory'] = df['Subcategory'].str.strip()
```

```python
df['Brand'].unique()
```

```python
df['Brand'] = df['Brand'].str.strip()
df['Brand'] = df['Brand'].str.capitalize()
```

```python
df['Brand'] = df['Brand'].str.strip()
```

```python
df['Stock_Quantity'].unique()
```

```python
df = df.dropna()
```

```python
df['Stock_Quantity'] = df['Stock_Quantity'].str.replace('pcs','')
```

```python
df['Stock_Quantity'] = df['Stock_Quantity'].astype(int)
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```
