```python
import pandas as pd
df = pd.read_csv('/content/Orders.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Order_Date'].unique()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed', dayfirst=True, errors='coerce')
```

```python
df['Order_Date'] = df['Order_Date'].dt.strftime('%d/%m/%Y')
```

```python
df.info()
```

```python
df.head()
```

```python
df['Order_Status'].unique()
```

```python
df['Order_Status'] = df['Order_Status'].str.strip()
df['Order_Status'] = df['Order_Status'].str.capitalize()
```

```python
df['Sales_Channel'].unique()
```

```python
df['Sales_Channel'] = df['Sales_Channel'].str.strip()
df['Sales_Channel'] = df['Sales_Channel'].str.capitalize()
```

```python
df= df.drop(['Shipping_Date','Delivery_Date'],axis = 1)
```

```python
df = df.dropna()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```
