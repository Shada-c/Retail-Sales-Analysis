```python
import pandas as pd
df = pd.read_csv('/content/OrderDetails.csv')
```

```python
df['Quantity'].unique()
```

```python
import numpy as np
df['Quantity']= df['Quantity'].replace('unknown',np.nan)
```

```python
df['Quantity']= df['Quantity'].replace('three',np.nan)
```

```python
df['Quantity'] = df['Quantity'].astype(str).str.replace(r'\s*(pcs|units)', '', regex=True)
```

```python
from sklearn.impute import SimpleImputer

si = SimpleImputer(strategy= 'mean')

df['Quantity'] = si.fit_transform(df[['Quantity']])
```

```python
df['Quantity'] = df['Quantity'].astype(int)
```

```python
df['Quantity'] = df['Quantity'].abs()
```

```python
df['Quantity'].describe()
```

```python
Q1 = df['Quantity'].quantile(0.25)
Q3 = df['Quantity'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Quantity'] < lower) | (df['Quantity'] > upper)]

print(outliers['Quantity'])
```

```python
df['Quantity'].value_counts()
```

```python
df = df[df['Quantity'].between(0, 100)]
```

```python
df['Price'].unique()
```

```python
df['Price']= df['Price'].replace('unknown',np.nan)
```

```python
df['Price'] = df['Price'].str.replace('₹','')
```

```python
df['Price'] = df['Price'].str.replace('INR','')
```

```python
df['Price'] = df['Price'].str.replace(',','')
```

```python
from sklearn.impute import SimpleImputer

si = SimpleImputer(strategy= 'median')

df['Price'] = si.fit_transform(df[['Price']])
```

```python
df['Price'] = df['Price'].astype(float)
```

```python
df.info()
```

```python
df['Unit_Cost'].unique()
```

```python
df['Unit_Cost']= df['Unit_Cost'].replace('unknown',np.nan)
```

```python
df = df.dropna()
```

```python
df['Unit_Cost'] = df['Unit_Cost'].str.replace('₹','')
```

```python
df['Unit_Cost'] = df['Unit_Cost'].str.replace('INR','')
```

```python
df['Unit_Cost'] = df['Unit_Cost'].str.replace(',','')
```

```python
df['Unit_Cost'] = df['Unit_Cost'].astype(float)
```

```python
df['Discount'].unique()
```

```python
df['Discount']= df['Discount'].replace('unknown',np.nan)
df['Discount'] = df['Discount'].str.replace('₹','')
df['Discount'] = df['Discount'].str.replace('INR','')
df['Discount'] = df['Discount'].str.replace(',','')
df['Discount'] = df['Discount'].astype(float)
```

```python
df['Line_Total'].unique()
```

```python
df['Line_Total']= df['Line_Total'].replace('unknown',np.nan)
df['Line_Total'] = df['Line_Total'].str.replace('₹','')
df['Line_Total'] = df['Line_Total'].str.replace('INR','')
df['Line_Total'] = df['Line_Total'].str.replace(',','')
df['Line_Total'] = df['Line_Total'].astype(float)
```

```python
df['Total_Cost']= df['Total_Cost'].replace('unknown',np.nan)
df['Total_Cost'] = df['Total_Cost'].str.replace('₹','')
df['Total_Cost'] = df['Total_Cost'].str.replace('INR','')
df['Total_Cost'] = df['Total_Cost'].str.replace(',','')
df['Total_Cost'] = df['Total_Cost'].astype(float)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df['Discount'] = df['Discount'].fillna('0')
```

```python
df.head()
```

```python
df['Line_Total'] = df['Line_Total'].fillna(df['Quantity'] * df['Price'] - df['Discount'])
```

```python
df.info()
```

```python
df['Total_Cost'] = df['Total_Cost'].fillna(df['Quantity'] * df['Unit_Cost'])
```

```python
df.to_csv('ORDER_DETAILS.csv', index=False)
```
